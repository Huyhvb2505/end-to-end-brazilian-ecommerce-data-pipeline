```python
import os
from pyspark.sql import SparkSession
from pyspark.sql.functions import *

spark = SparkSession.builder \
        .appName("Brazilian-Ecommerce-load-Gold") \
        .master("spark://spark-master:7077") \
        .config("spark.sql.adaptive.enabled", "true") \
        .config("spark.sql.adaptive.coalescePartitions.enabled", "true") \
        .config("spark.jars.packages", "org.postgresql:postgresql:42.6.0,org.apache.hadoop:hadoop-aws:3.3.4") \
        .config("spark.hadoop.fs.s3a.endpoint", "http://minio:9000") \
        .config("spark.hadoop.fs.s3a.access.key", "minio_admin") \
        .config("spark.hadoop.fs.s3a.secret.key", "minio_secure_pwd") \
        .config("spark.hadoop.fs.s3a.path.style.access", "true") \
        .config("spark.hadoop.fs.s3a.impl", "org.apache.hadoop.fs.s3a.S3AFileSystem") \
        .config("spark.hadoop.fs.s3a.connection.ssl.enabled", "false") \
        .config("spark.executor.memory", "1g") \
        .config("spark.driver.memory", "1g") \
        .getOrCreate()

# set log level
spark.sparkContext.setLogLevel("WARN")

print("✅ Spark Session created successfully")
print(f"📊 Spark Version: {spark.version}")
print(f"🔧 Application Name: {spark.sparkContext.appName}")
print(f"🎯 Master URL: {spark.sparkContext.master}")
```

```python
# 1. ĐỌC DỮ LIỆU TỪ TẦNG SILVER (MINIO)


silver_path = "s3a://processed-data/"

df_products_silver = spark.read.parquet(f"{silver_path}products")
df_translation_silver = spark.read.parquet(f"{silver_path}category_name_translation")
df_orders_silver = spark.read.parquet(f"{silver_path}orders")
df_order_items_silver = spark.read.parquet(f"{silver_path}order_items")
df_customers_silver = spark.read.parquet(f"{silver_path}customers")
df_sellers_silver = spark.read.parquet(f"{silver_path}sellers")
df_payments_silver = spark.read.parquet(f"{silver_path}payments")


dim_product = df_products_silver.join(
    df_translation_silver, 
    on="product_category_name", 
    how="left"   
).withColumnRenamed("product_category_name_english", "category_name_en") \
 .select("product_id", "category_name_en") \
 .withColumn("product_key", monotonically_increasing_id())

df_date_range = spark.sql("SELECT sequence(to_date('2016-09-04'), to_date('2018-10-18'), interval 1 day) as date_array")
dim_date = df_date_range.select(explode(col("date_array")).alias("full_date"))
dim_date = dim_date.withColumn("date_key", date_format(col("full_date") , "yyyyMMdd").cast("int")) \
                   .withColumn("month", month(col("full_date"))) \
                   .withColumn("quarter", quarter(col("full_date"))) \
                   .withColumn("year", year(col("full_date"))) \
                   .select("date_key", "full_date", "month", "quarter", "year")

dim_customer = df_customers_silver.select(
    "customer_id", "customer_unique_id", "customer_city", "customer_state"
).withColumn("customer_key", monotonically_increasing_id())

dim_seller = df_sellers_silver.select(
    "seller_id", "seller_state"
).withColumn("seller_key", monotonically_increasing_id())

# ==========================================
# 3. TRANSFORM & TẠO FACT TABLES (GOLD)
# ==========================================
fact_orders = df_orders_silver \
    .withColumn("delivery_lead_time_days", datediff(col("order_delivered_customer_date") , col("order_purchase_timestamp"))) \
    .withColumn("is_late",
                when(col("order_delivered_customer_date") > col("order_estimated_delivery_date"), True)
                .when(col("order_delivered_customer_date").isNull(), None)
                .otherwise(False)) \
    .withColumn("date_key", date_format(col("order_purchase_timestamp"), "yyyyMMdd").cast("int")) \
    .join(dim_customer.select("customer_id", "customer_key"), on="customer_id", how="left") \
    .select("customer_key","date_key","order_id", "order_status", "delivery_lead_time_days", "is_late")

fact_order_items = df_order_items_silver \
    .withColumn("item_total_value", col("price") + col("freight_value")) \
    .join(dim_product.select("product_id", "product_key"), on="product_id", how="left") \
    .join(dim_seller.select("seller_id", "seller_key"), on="seller_id", how="left") \
    .join(fact_orders.select("order_id", "date_key"), on="order_id", how="left") \
    .select("product_key", "seller_key", "date_key","order_id", "order_item_id",  "price", "freight_value", "item_total_value")

fact_payment = df_payments_silver \
    .join(fact_orders.select("order_id", "date_key"), on="order_id", how="left") \
    .select("date_key","order_id", "payment_sequential",  "payment_type", "payment_value")

# ==========================================
# 4. NẠP DỮ LIỆU VÀO POSTGRESQL
# ==========================================
db_url = "jdbc:postgresql://postgres-dw:5432/brazilian_ecommerce_dw"
db_properties = {
    "user": "de_user",
    "password": "de_secure_password_2026",
    "driver": "org.postgresql.Driver"
}

def load_to_db(df, table_name):
    print(f"📥 Đang nạp {table_name} vào PostgreSQL...")
    df.write.jdbc(url=db_url, table=table_name, mode="append", properties=db_properties)
    print(f"✅ Hoàn tất: {table_name}")

# Nạp Dimension trước để đảm bảo tính toàn vẹn khóa ngoại (Foreign Keys)
load_to_db(dim_customer, "dim_customer")
load_to_db(dim_product, "dim_product")
load_to_db(dim_seller, "dim_seller")
load_to_db(dim_date, "dim_date")

# Nạp Fact sau
load_to_db(fact_orders, "fact_orders")
load_to_db(fact_order_items, "fact_order_items")
load_to_db(fact_payment, "fact_payment")

print("\n🚀 TOÀN BỘ PIPELINE ĐÃ CHẠY THÀNH CÔNG!")
```

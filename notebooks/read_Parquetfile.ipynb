```python
import pandas as pd

# Cấu hình thông tin kết nối MinIO dưới dạng dictionary cho s3fs
storage_options = {
    "key": "minio_admin",  # MinIO Access Key
    "secret": "minio_secure_pwd",  # MinIO Secret Key
    "client_kwargs": {
        "endpoint_url": "http://localhost:9001"
    },  # Endpoint của MinIO (đổi sang http://... nếu chạy local)
}

# Đường dẫn S3 tới file trên MinIO (định dạng: s3://tên_bucket/đường_dẫn_file)
s3_path = "s3://raw-data/olist_customers_dataset.parquet"

# Đọc trực tiếp vào DataFrame
df = pd.read_parquet(s3_path, storage_options=storage_options)

# Xem thử dữ liệu
print(df.head())
print(f"Tổng số dòng: {len(df)}")
```

```python
import boto3

s3 = boto3.client(
    's3',
    endpoint_url='http://localhost:9000',
    aws_access_key_id='minio_admin',
    aws_secret_access_key='minio_secure_pwd',
)

# Đọc nội dung file từ bucket
response = s3.get_object(Bucket='raw-data', Key='s3://raw-data/olist_customers_dataset.csv')
content = response['Body'].read().decode('utf-8')
print(content)
```
